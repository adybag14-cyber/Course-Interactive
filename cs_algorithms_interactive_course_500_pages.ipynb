````python

from collections import deque, defaultdict, OrderedDict
from functools import cache
import heapq, math, random

# Shared algorithm laboratory utilities used throughout the notebook.

def lower_bound(a, target):
    lo, hi = 0, len(a)
    while lo < hi:
        mid = (lo + hi) // 2
        if a[mid] < target:
            lo = mid + 1
        else:
            hi = mid
    return lo

def merge_sort(a):
    if len(a) <= 1:
        return list(a)
    mid = len(a) // 2
    left = merge_sort(a[:mid])
    right = merge_sort(a[mid:])
    out = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            out.append(left[i]); i += 1
        else:
            out.append(right[j]); j += 1
    return out + left[i:] + right[j:]

def bfs(graph, start):
    seen = {start}
    q = deque([start])
    order = []
    while q:
        u = q.popleft()
        order.append(u)
        for v in graph[u]:
            if v not in seen:
                seen.add(v)
                q.append(v)
    return order

class DSU:
    def __init__(self, n):
        self.parent = list(range(n))
        self.size = [1] * n
    def find(self, x):
        while x != self.parent[x]:
            self.parent[x] = self.parent[self.parent[x]]
            x = self.parent[x]
        return x
    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False
        if self.size[ra] < self.size[rb]:
            ra, rb = rb, ra
        self.parent[rb] = ra
        self.size[ra] += self.size[rb]
        return True

def dijkstra(graph, source):
    dist = {source: 0}
    parent = {source: None}
    pq = [(0, source)]
    while pq:
        d, u = heapq.heappop(pq)
        if d != dist.get(u):
            continue
        for v, w in graph[u]:
            nd = d + w
            if nd < dist.get(v, float('inf')):
                dist[v] = nd
                parent[v] = u
                heapq.heappush(pq, (nd, v))
    return dist, parent

def knapsack(weights, values, capacity):
    dp = [0] * (capacity + 1)
    for w, val in zip(weights, values):
        for c in range(capacity, w - 1, -1):
            dp[c] = max(dp[c], dp[c - w] + val)
    return dp[capacity]

def prefix_function(pattern):
    pi = [0] * len(pattern)
    j = 0
    for i in range(1, len(pattern)):
        while j and pattern[i] != pattern[j]:
            j = pi[j - 1]
        if pattern[i] == pattern[j]:
            j += 1
        pi[i] = j
    return pi

def kmp_search(text, pattern):
    if not pattern:
        return list(range(len(text) + 1))
    pi, j, hits = prefix_function(pattern), 0, []
    for i, ch in enumerate(text):
        while j and ch != pattern[j]:
            j = pi[j - 1]
        if ch == pattern[j]:
            j += 1
        if j == len(pattern):
            hits.append(i - j + 1)
            j = pi[j - 1]
    return hits

def orientation(a, b, c):
    return (b[0] - a[0]) * (c[1] - a[1]) - (b[1] - a[1]) * (c[0] - a[0])

def convex_hull(points):
    pts = sorted(set(points))
    if len(pts) <= 1:
        return pts
    lower = []
    for p in pts:
        while len(lower) >= 2 and orientation(lower[-2], lower[-1], p) <= 0:
            lower.pop()
        lower.append(p)
    upper = []
    for p in reversed(pts):
        while len(upper) >= 2 and orientation(upper[-2], upper[-1], p) <= 0:
            upper.pop()
        upper.append(p)
    return lower[:-1] + upper[:-1]

print('Algorithm lab loaded. Try lower_bound([1,3,3,8], 3), bfs(...), dijkstra(...), or kmp_search(...).')
````

## Reference implementation sketch

```python
def interval_scheduling(intervals):
    chosen = []
    current_end = float('-inf')
    for start, end in sorted(intervals, key=lambda x: x[1]):
        if start >= current_end:
            chosen.append((start, end))
            current_end = end
    return chosen
```

```python
def reservoir_sample(stream, k):
    sample = []
    for i, item in enumerate(stream, start=1):
        if i <= k:
            sample.append(item)
        else:
            j = random.randrange(i)
            if j < k:
```

**Chapter 4: Python and JavaScript Algorithm Tooling**  
**Angle:** optimization

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(V + E) | O(V) | visited-state correctness |

## Working notes
- Optimize only after naming the bottleneck. Replacing a scan with a heap, hash table, prefix summary, or monotone structure should change a measurable line in the cost model.
- Key risk: visited-state correctness. Make this risk visible in tests and in code comments.

**Common pitfall:** Relying on a library without reading its ordering, mutation, or thread-safety contract can silently invalidate an algorithm.

## Reference implementation sketch

```python
from collections import deque, defaultdict

def topo_sort(edges):
    graph, indeg = defaultdict(list), defaultdict(int)
    nodes = set()
    for u, v in edges:
        graph[u].append(v); indeg[v] += 1
        nodes.update([u, v])
    q = deque([u for u in nodes if indeg[u] == 0])
    order = []
    while q:
        u = q.popleft(); order.append(u)
        for v in graph[u]:
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)
    if len(order) != len(nodes):
        raise ValueError('cycle detected')
    return order
```

# Page 051 — Static arrays and memory locality

**Chapter 5: Arrays, Strings, and Memory**  
**Angle:** implementation

Arrays give constant-time indexing and excellent locality. Many linear-time algorithms are just careful scans with a small amount of maintained state.

## Core use
Use arrays when random access, contiguous traversal, prefix summaries, sliding windows, or in-place partitioning dominate the workload.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| often O(n) | O(1) to O(n) | off-by-one and resizing |

## Working notes
- Start with the smallest representation that preserves the invariant; extra fields should pay for themselves by reducing transition cost or simplifying proof.
- Key risk: off-by-one and resizing. Make this risk visible in tests and in code comments.

**Common pitfall:** A logically simple operation such as insertion at the front of an array can be linear because many elements must move.

## Reference implementation sketch

```python
def prefix_function(pattern):
    pi = [0] * len(pattern)
    j = 0
    for i in range(1, len(pattern)):
        while j and pattern[i] != pattern[j]:
            j = pi[j - 1]
        if pattern[i] == pattern[j]:
            j += 1
        pi[i] = j
    return pi

def kmp_search(text, pattern):
    if not pattern:
        return list(range(len(text) + 1))
    pi, j, hits = prefix_function(pattern), 0, []
    for i, ch in enumerate(text):
        while j and ch != pattern[j]:
            j = pi[j - 1]
        if ch == pattern[j]:
            j += 1
        if j == len(pattern):
            hits.append(i - j + 1)
            j = pi[j - 1]
    return hits
```

**Question:** What single assumption makes the usual Static arrays and memory locality analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time often O(n) and space O(1) to O(n), but the exact assumption is the one stated in the page notes.

```python
def quicksort(a):
    if len(a) <= 1:
        return a[:]
    pivot = a[len(a) // 2]
    less  = [x for x in a if x < pivot]
    equal = [x for x in a if x == pivot]
    more  = [x for x in a if x > pivot]
    return quicksort(less) + equal + quicksort(more)
```

# Page 079 — Sorting problem model

**Chapter 7: Sorting Algorithms**  
**Angle:** practice

Sorting imposes order so later operations become simpler. Comparison sorting has a lower bound, while non-comparison sorts exploit restricted keys.

## Core use
Sort when order enables scanning, deduplication, grouping, binary search, greedy selection, or sweep-line processing.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(n log n) | O(n) or O(log n) | stability and comparator |

## Working notes
- Practice by inventing a tiny input, tracing every step, then writing a randomized checker that compares against a slow brute-force implementation.
- Key risk: stability and comparator. Make this risk visible in tests and in code comments.

**Common pitfall:** A comparator that is not transitive can make even a correct sorting implementation produce surprising results.

## Reference implementation sketch

```python
def merge_sort(a):
    if len(a) <= 1:
        return a[:]
    mid = len(a) // 2
    left = merge_sort(a[:mid])
    right = merge_sort(a[mid:])
    out, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            out.append(left[i]); i += 1
        else:
            out.append(right[j]); j += 1
    return out + left[i:] + right[j:]
```

## Checkpoint quiz
**Question:** What single assumption makes the usual Sorting problem model analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time O(n log n) and space O(n) or O(log n), but the exact assumption is the one stated in the page notes.

# Page 101 — Closest pair divide-and-conquer

**Chapter 8: Recursion and Divide and Conquer**  
**Angle:** implementation

Divide-and-conquer solves a problem by splitting it into smaller subproblems, solving those recursively, and combining the results.

## Core use
Use it when the problem decomposes cleanly, the combine step is efficient, or subproblems can run independently.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| often recurrence-based | recursion depth | missing base cases |

## Working notes
- Start with the smallest representation that preserves the invariant; extra fields should pay for themselves by reducing transition cost or simplifying proof.
- Key risk: missing base cases. Make this risk visible in tests and in code comments.

**Common pitfall:** A recursive solution without a shrinking argument may look elegant while never terminating.

## Reference implementation sketch

```python
def orientation(a, b, c):
    # positive: counterclockwise, negative: clockwise, zero: collinear
    return (b[0] - a[0]) * (c[1] - a[1]) - (b[1] - a[1]) * (c[0] - a[0])

def convex_hull(points):
    pts = sorted(set(points))
    if len(pts) <= 1:
        return pts
    lower = []
    for p in pts:
        while len(lower) >= 2 and orientation(lower[-2], lower[-1], p) <= 0:
            lower.pop()
        lower.append(p)
    upper = []
    for p in reversed(pts):
        while len(upper) >= 2 and orientation(upper[-2], upper[-1], p) <= 0:
            upper.pop()
        upper.append(p)
    return lower[:-1] + upper[:-1]
```

**Answer:** The representation and invariant must match the intended model; the common cost summary is time often recurrence-based and space recursion depth, but the exact assumption is the one stated in the page notes.

```python
def quicksort(a):

    pivot = a[len(a) // 2]
    less  = [x for x in a if x < pivot]
    equal = [x for x in a if x == pivot]
```

## Complexity lens

## Working notes
- A proof usually follows the implementation structure: initialization establishes the invariant, each step preserves it, and termination converts it into the postcondition.

## Reference implementation sketch

```python
def permutations(items):
    items = list(items)
    used = [False] * len(items)
    out = []
    def search(path):
        if len(path) == len(items):
            out.append(tuple(path)); return
        for i, x in enumerate(items):
            if not used[i]:
                used[i] = True
                path.append(x)
                search(path)
                path.pop()
                used[i] = False
    search([])
    return out
```

# Page 111 — Stack ADT

**Chapter 9: Stacks, Queues, Deques, and Linked Lists**  
**Angle:** implementation

Linear ADTs control the order in which work is processed. Stacks give last-in-first-out behavior, queues give first-in-first-out behavior, and deques support both ends.

## Core use
Use these structures for traversal frontiers, undo history, parsing, sliding windows, and local neighbor relationships.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(1) core operations | O(n) storage | wrong end operations |

## Working notes
- Start with the smallest representation that preserves the invariant; extra fields should pay for themselves by reducing transition cost or simplifying proof.
- Key risk: wrong end operations. Make this risk visible in tests and in code comments.

**Common pitfall:** Using a structure that supports the right operation at the wrong end can change a linear traversal into a quadratic one.

## Reference implementation sketch

```python
from collections import deque

def bfs(graph, start):
    seen = {start}
    q = deque([start])
    order = []
    while q:
        u = q.popleft()
        order.append(u)
        for v in graph[u]:
            if v not in seen:
                seen.add(v)
                q.append(v)
    return order
```

```python
def dfs(graph, start):
    seen, order = set(), []
    def visit(u):
        seen.add(u)
        order.append(u)
```

**Angle:** proof idea

Trees encode hierarchy. Traversals turn recursive structure into a sequence, and augmentation lets each node summarize its subtree.

## Core use
Use trees for hierarchical data, ordered maps, range queries, parse trees, prefix dictionaries, and divide-and-conquer state.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(n) traversal | O(h) stack | unbalanced height |

## Working notes
- A proof usually follows the implementation structure: initialization establishes the invariant, each step preserves it, and termination converts it into the postcondition.
- Key risk: unbalanced height. Make this risk visible in tests and in code comments.

**Common pitfall:** Confusing node count n with height h leads to wrong bounds for recursive tree operations.

## Reference implementation sketch

```python
class TrieNode(dict):
    def __init__(self):
        super().__init__()
        self.end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()
    def add(self, word):
        node = self.root
        for ch in word:
            node = node.setdefault(ch, TrieNode())
        node.end = True
    def contains(self, word):
        node = self.root
        for ch in word:
            if ch not in node:
                return False
            node = node[ch]
        return node.end
```

# Page 159 — BST invariant

**Chapter 12: Binary Search Trees and Balanced Trees**  
**Angle:** practice

Search trees maintain an order invariant so lookup follows one path instead of scanning all keys. Balancing keeps that path short.

## Core use
Use balanced trees when you need ordered operations such as predecessor, successor, rank, range query, or sorted iteration under updates.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(log n) | O(1) | requires monotone predicate |

## Working notes
- Practice by inventing a tiny input, tracing every step, then writing a randomized checker that compares against a slow brute-force implementation.
- Key risk: requires monotone predicate. Make this risk visible in tests and in code comments.

**Common pitfall:** A plain BST can become a linked list if keys arrive in sorted order.

## Reference implementation sketch

```python
def lower_bound(a, target):
    lo, hi = 0, len(a)
    while lo < hi:
        mid = (lo + hi) // 2
        if a[mid] < target:
            lo = mid + 1
        else:
            hi = mid
    return lo
```

# Page 175 — Heap invariant

**Chapter 13: Heaps and Priority Queues**  
**Angle:** proof idea

A heap is a partial order that keeps the best item at the root while avoiding the cost of fully sorting every element.

## Core use
Use heaps for repeated minimum/maximum extraction, top-k, event simulation, k-way merge, and graph algorithms with tentative priorities.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| push/pop O(log n), peek O(1) | O(n) | stale priorities |

## Working notes
- A proof usually follows the implementation structure: initialization establishes the invariant, each step preserves it, and termination converts it into the postcondition.
- Key risk: stale priorities. Make this risk visible in tests and in code comments.

**Common pitfall:** Priority queues usually do not support efficient arbitrary priority decrease unless you add an index or use lazy deletion.

## Reference implementation sketch

```python
import heapq

def k_smallest(items, k):
    heap = list(items)
    heapq.heapify(heap)
    return [heapq.heappop(heap) for _ in range(min(k, len(heap)))]
```

## Checkpoint quiz
**Question:** What single assumption makes the usual Heap invariant analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time push/pop O(log n), peek O(1) and space O(n), but the exact assumption is the one stated in the page notes.

## Practice task

# Page 189 — DSU parent forest

**Chapter 14: Disjoint Set Union**  
**Angle:** practice

DSU maintains a partition of elements into disjoint sets. It answers which set contains an element and merges sets efficiently.

## Core use
Use DSU for incremental connectivity, Kruskal's MST, grouping equalities, offline connectivity, and image/component labeling.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| near-constant amortized | O(n) | path compression / union size |

## Working notes
- Practice by inventing a tiny input, tracing every step, then writing a randomized checker that compares against a slow brute-force implementation.
- Key risk: path compression / union size. Make this risk visible in tests and in code comments.

**Common pitfall:** Forgetting path compression or union by size can turn practical near-constant operations into tall-tree traversals.

## Reference implementation sketch

```python
class DSU:
    def __init__(self, n):
        self.parent = list(range(n))
        self.size = [1] * n
    def find(self, x):
        while x != self.parent[x]:
            self.parent[x] = self.parent[self.parent[x]]
            x = self.parent[x]
        return x
    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False
        if self.size[ra] < self.size[rb]:
            ra, rb = rb, ra
        self.parent[rb] = ra
        self.size[ra] += self.size[rb]
        return True
```

## Checkpoint quiz
**Question:** What single assumption makes the usual DSU parent forest analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time near-constant amortized and space O(n), but the exact assumption is the one stated in the page notes.

# Page 199 — Graph representations

**Chapter 15: Graph Basics and Traversal**  
**Angle:** practice

Graphs model relationships. Traversal algorithms explore reachable states while preventing repeated work with a visited set.

## Core use
Use BFS for layer-by-layer shortest paths in unweighted graphs and DFS for structure such as components, cycles, bridges, and finish times.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(V+E) | O(V+E) | visited-state mistakes |

## Working notes
- Practice by inventing a tiny input, tracing every step, then writing a randomized checker that compares against a slow brute-force implementation.
- Key risk: visited-state mistakes. Make this risk visible in tests and in code comments.

**Common pitfall:** Marking nodes visited too late can enqueue or recurse into the same state many times.

## Reference implementation sketch

```python
from collections import deque

def bfs(graph, start):
    seen = {start}
    q = deque([start])
    order = []
    while q:
        u = q.popleft()
        order.append(u)
        for v in graph[u]:
            if v not in seen:
                seen.add(v)
                q.append(v)
    return order
```

## Checkpoint quiz
**Question:** What single assumption makes the usual Graph representations analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time O(V+E) and space O(V+E), but the exact assumption is the one stated in the page notes.

# Page 217 — DAG model

**Chapter 16: DAGs and Topological Algorithms**  
**Angle:** optimization

A directed acyclic graph has at least one ordering where every edge points forward. This order turns dependency problems into linear scans.

## Core use
Use topological order for build systems, course scheduling, critical paths, DAG shortest paths, and DP over dependencies.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(V + E) | O(V) | visited-state correctness |

## Working notes
- Optimize only after naming the bottleneck. Replacing a scan with a heap, hash table, prefix summary, or monotone structure should change a measurable line in the cost model.
- Key risk: visited-state correctness. Make this risk visible in tests and in code comments.

**Common pitfall:** If a cycle exists, no topological order exists; silently producing a partial order is a bug.

## Reference implementation sketch

```python
from collections import deque, defaultdict

def topo_sort(edges):
    graph, indeg = defaultdict(list), defaultdict(int)
    nodes = set()
    for u, v in edges:
        graph[u].append(v); indeg[v] += 1
        nodes.update([u, v])
    q = deque([u for u in nodes if indeg[u] == 0])
    order = []
    while q:
        u = q.popleft(); order.append(u)
        for v in graph[u]:
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)
    if len(order) != len(nodes):
        raise ValueError('cycle detected')
    return order
```

## Checkpoint quiz
**Question:** What single assumption makes the usual DAG model analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time O(V + E) and space O(V), but the exact assumption is the one stated in the page notes.

# Page 221 — DAG shortest paths

**Chapter 16: DAGs and Topological Algorithms**  
**Angle:** implementation

A directed acyclic graph has at least one ordering where every edge points forward. This order turns dependency problems into linear scans.

## Core use
Use topological order for build systems, course scheduling, critical paths, DAG shortest paths, and DP over dependencies.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(V + E) | O(V) | visited-state correctness |

## Working notes
- Start with the smallest representation that preserves the invariant; extra fields should pay for themselves by reducing transition cost or simplifying proof.
- Key risk: visited-state correctness. Make this risk visible in tests and in code comments.

**Common pitfall:** If a cycle exists, no topological order exists; silently producing a partial order is a bug.

## Reference implementation sketch

```python
import heapq

def dijkstra(graph, source):
    dist = {source: 0}
    parent = {source: None}
    pq = [(0, source)]
    while pq:
        d, u = heapq.heappop(pq)
        if d != dist.get(u):
            continue
        for v, w in graph[u]:
            nd = d + w
            if nd < dist.get(v, float('inf')):
                dist[v] = nd
                parent[v] = u
                heapq.heappush(pq, (nd, v))
    return dist, parent
```

## Checkpoint quiz
**Question:** What single assumption makes the usual DAG shortest paths analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time O(V + E) and space O(V), but the exact assumption is the one stated in the page notes.

# Page 222 — DAG longest paths

**Chapter 16: DAGs and Topological Algorithms**  
**Angle:** invariants

A directed acyclic graph has at least one ordering where every edge points forward. This order turns dependency problems into linear scans.

## Core use
Use topological order for build systems, course scheduling, critical paths, DAG shortest paths, and DP over dependencies.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(V + E) | O(V) | visited-state correctness |

## Working notes
- Write the invariant in a sentence before code. At every step, check what remains true, what shrinks, and what answer information is already final.
- Key risk: visited-state correctness. Make this risk visible in tests and in code comments.

**Common pitfall:** If a cycle exists, no topological order exists; silently producing a partial order is a bug.

## Reference implementation sketch

```python
from collections import deque, defaultdict

def topo_sort(edges):
    graph, indeg = defaultdict(list), defaultdict(int)
    nodes = set()
    for u, v in edges:
        graph[u].append(v); indeg[v] += 1
        nodes.update([u, v])
    q = deque([u for u in nodes if indeg[u] == 0])
    order = []
    while q:
        u = q.popleft(); order.append(u)
        for v in graph[u]:
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)
    if len(order) != len(nodes):
        raise ValueError('cycle detected')
    return order
```

## Checkpoint quiz
**Question:** What single assumption makes the usual DAG longest paths analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time O(V + E) and space O(V), but the exact assumption is the one stated in the page notes.

```python
def k_smallest(items, k):
```

# Page 236 — Floyd-Warshall

**Chapter 17: Shortest Paths**  
**Angle:** debugging

Shortest-path algorithms maintain distance estimates and relax edges until no better path remains under the algorithm's assumptions.

## Core use
Use BFS for unweighted graphs, Dijkstra for nonnegative weights, Bellman-Ford for negative edges, and Floyd-Warshall for dense all-pairs problems.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(V^3) | O(V^2) | dense all-pairs |

## Working notes
- Log the state that defines progress, not every variable. For graph algorithms this is usually frontier/visited; for DP it is the current state and dependencies.
- Key risk: dense all-pairs. Make this risk visible in tests and in code comments.

**Common pitfall:** Dijkstra's greedy choice is not valid in the presence of negative edge weights.

## Reference implementation sketch

```python
import heapq

def dijkstra(graph, source):
    dist = {source: 0}
    parent = {source: None}
    pq = [(0, source)]
    while pq:
        d, u = heapq.heappop(pq)
        if d != dist.get(u):
            continue
        for v, w in graph[u]:
            nd = d + w
            if nd < dist.get(v, float('inf')):
                dist[v] = nd
                parent[v] = u
                heapq.heappush(pq, (nd, v))
    return dist, parent
```

# Page 247 — Kruskal's algorithm

**Chapter 18: Spanning Trees and Connectivity**  
**Angle:** optimization

A spanning tree connects all vertices with no cycles. MST algorithms choose edges that minimize total weight while preserving connectivity.

## Core use
Use MSTs for network design, clustering, approximation, and as a lens for cut/cycle reasoning.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(E log V) common | O(V+E) | disconnected inputs |

## Working notes
- Optimize only after naming the bottleneck. Replacing a scan with a heap, hash table, prefix summary, or monotone structure should change a measurable line in the cost model.
- Key risk: disconnected inputs. Make this risk visible in tests and in code comments.

**Common pitfall:** An MST only exists for a connected graph; disconnected graphs produce a minimum spanning forest.

## Reference implementation sketch

```python
class DSU:
    def __init__(self, n):
        self.parent = list(range(n))
        self.size = [1] * n
    def find(self, x):
        while x != self.parent[x]:
            self.parent[x] = self.parent[self.parent[x]]
            x = self.parent[x]
        return x
    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False
        if self.size[ra] < self.size[rb]:
            ra, rb = rb, ra
        self.parent[rb] = ra
        self.size[ra] += self.size[rb]
        return True
```

## Checkpoint quiz
**Question:** What single assumption makes the usual Kruskal&#x27;s algorithm analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time O(E log V) common and space O(V+E), but the exact assumption is the one stated in the page notes.

# Page 253 — Dynamic connectivity

**Chapter 18: Spanning Trees and Connectivity**  
**Angle:** complexity

A spanning tree connects all vertices with no cycles. MST algorithms choose edges that minimize total weight while preserving connectivity.

## Core use
Use MSTs for network design, clustering, approximation, and as a lens for cut/cycle reasoning.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(E log V) common | O(V+E) | disconnected inputs |

## Working notes
- Budget the cost as O(E log V) common time and O(V+E) space in the common model, then revisit the bound when the input representation changes.
- Key risk: disconnected inputs. Make this risk visible in tests and in code comments.

**Common pitfall:** An MST only exists for a connected graph; disconnected graphs produce a minimum spanning forest.

## Reference implementation sketch

```python
def kruskal(n, edges):
    dsu = DSU(n)
    total, chosen = 0, []
    for w, u, v in sorted(edges):
        if dsu.union(u, v):
            total += w
            chosen.append((u, v, w))
    if len(chosen) != n - 1:
        raise ValueError('graph is disconnected')
    return total, chosen
```

**Chapter 18: Spanning Trees and Connectivity**  
**Angle:** proof idea

A spanning tree connects all vertices with no cycles. MST algorithms choose edges that minimize total weight while preserving connectivity.

## Core use
Use MSTs for network design, clustering, approximation, and as a lens for cut/cycle reasoning.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(V + E) | O(V) | visited-state correctness |

## Working notes
- A proof usually follows the implementation structure: initialization establishes the invariant, each step preserves it, and termination converts it into the postcondition.
- Key risk: visited-state correctness. Make this risk visible in tests and in code comments.

**Common pitfall:** An MST only exists for a connected graph; disconnected graphs produce a minimum spanning forest.

## Reference implementation sketch

```python
def dfs(graph, start):
    seen, order = set(), []
    def visit(u):
        seen.add(u)
        order.append(u)
        for v in graph[u]:
            if v not in seen:
                visit(v)
    visit(start)
    return order
```

# Page 276 — DAG view of DP

**Chapter 20: Dynamic Programming Foundations**  
**Angle:** debugging

Dynamic programming caches overlapping subproblems and evaluates states in an order that respects dependencies.

## Core use
Use DP when optimal solutions combine optimal subsolutions and the same subproblems appear many times.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| states × transition cost | state table | state definition |

## Working notes
- Log the state that defines progress, not every variable. For graph algorithms this is usually frontier/visited; for DP it is the current state and dependencies.
- Key risk: state definition. Make this risk visible in tests and in code comments.

**Common pitfall:** A DP with a vague state often leads to duplicate states, missing information, or an exponential transition loop.

## Reference implementation sketch

```python
from collections import deque, defaultdict

def topo_sort(edges):
    graph, indeg = defaultdict(list), defaultdict(int)
    nodes = set()
    for u, v in edges:
        graph[u].append(v); indeg[v] += 1
        nodes.update([u, v])
    q = deque([u for u in nodes if indeg[u] == 0])
    order = []
    while q:
        u = q.popleft(); order.append(u)
        for v in graph[u]:
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)
    if len(order) != len(nodes):
        raise ValueError('cycle detected')
    return order
```

## Checkpoint quiz
**Question:** What single assumption makes the usual DAG view of DP analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time states × transition cost and space state table, but the exact assumption is the one stated in the page notes.

# Page 298 — Partition DP

**Chapter 21: Sequence and String Dynamic Programming**  
**Angle:** applications

Sequence DP compares prefixes, suffixes, or intervals. The state usually names how much of each sequence has already been solved.

## Core use
Use it for alignment, edit distance, subsequence optimization, knapsack families, palindromes, and parsing-like problems.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| states × transition cost | state table | state definition |

## Working notes
- Look for Partition DP whenever data has structure: order, hierarchy, graph connectivity, repeated subproblems, locality, or constraints that rule out most candidates.
- Key risk: state definition. Make this risk visible in tests and in code comments.

**Common pitfall:** Filling the table in an order that uses uncomputed neighbors gives plausible but wrong values.

## Reference implementation sketch

```python
def quicksort(a):
    if len(a) <= 1:
        return a[:]
    pivot = a[len(a) // 2]
    less  = [x for x in a if x < pivot]
    equal = [x for x in a if x == pivot]
    more  = [x for x in a if x > pivot]
    return quicksort(less) + equal + quicksort(more)
```

# Page 302 — Optimal binary search tree

**Chapter 21: Sequence and String Dynamic Programming**  
**Angle:** invariants

Sequence DP compares prefixes, suffixes, or intervals. The state usually names how much of each sequence has already been solved.

## Core use
Use it for alignment, edit distance, subsequence optimization, knapsack families, palindromes, and parsing-like problems.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(log n) | O(1) | requires monotone predicate |

## Working notes
- Write the invariant in a sentence before code. At every step, check what remains true, what shrinks, and what answer information is already final.
- Key risk: requires monotone predicate. Make this risk visible in tests and in code comments.

**Common pitfall:** Filling the table in an order that uses uncomputed neighbors gives plausible but wrong values.

## Reference implementation sketch

```python
def lower_bound(a, target):
    lo, hi = 0, len(a)
    while lo < hi:
        mid = (lo + hi) // 2
        if a[mid] < target:
            lo = mid + 1
        else:
            hi = mid
    return lo
```

# Page 325 — Greedy choice property

**Angle:** proof idea

Greedy algorithms make locally optimal choices and rely on a structure theorem that those choices can be extended to a global optimum.

## Core use

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| often O(n log n) | often O(1)-O(n) | unproved local choices |

## Working notes
- A proof usually follows the implementation structure: initialization establishes the invariant, each step preserves it, and termination converts it into the postcondition.
- Key risk: unproved local choices. Make this risk visible in tests and in code comments.

**Common pitfall:** Greedy code is easy to write and easy to be wrong; proof is the algorithm, not decoration.

## Reference implementation sketch

```python
def interval_scheduling(intervals):
    chosen = []
    current_end = float('-inf')
    for start, end in sorted(intervals, key=lambda x: x[1]):
        if start >= current_end:
            chosen.append((start, end))
            current_end = end
    return chosen
```

# Page 328 — Fractional knapsack

**Chapter 23: Greedy Algorithms**  
**Angle:** applications

Greedy algorithms make locally optimal choices and rely on a structure theorem that those choices can be extended to a global optimum.

## Core use
Use greedy when an exchange argument, matroid-like structure, cut property, or stays-ahead proof is available.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| states × transition cost | state table | state definition |

## Working notes
- Look for Fractional knapsack whenever data has structure: order, hierarchy, graph connectivity, repeated subproblems, locality, or constraints that rule out most candidates.
- Key risk: state definition. Make this risk visible in tests and in code comments.

**Common pitfall:** Greedy code is easy to write and easy to be wrong; proof is the algorithm, not decoration.

## Reference implementation sketch

```python
def knapsack(weights, values, capacity):
    dp = [0] * (capacity + 1)
    for w, val in zip(weights, values):
        for c in range(capacity, w - 1, -1):
            dp[c] = max(dp[c], dp[c - w] + val)
    return dp[capacity]
```

```python
def has_pair_with_sum(a, target):
    a = sorted(a)
    lo, hi = 0, len(a) - 1
    while lo < hi:
        s = a[lo] + a[hi]
        if s == target:
            return True
        if s < target:
            lo += 1
        else:
            hi -= 1
    return False
```

# Page 341 — Backtracking template

**Chapter 24: Backtracking and Combinatorial Search**  
**Angle:** implementation

Backtracking explores a decision tree and abandons branches that cannot lead to valid or better solutions.

## Core use
Use it for permutations, constraint solving, puzzles, exact cover, game trees, and problems where pruning is powerful.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| exponential worst case | depth-dependent | pruning |

## Working notes
- Start with the smallest representation that preserves the invariant; extra fields should pay for themselves by reducing transition cost or simplifying proof.
- Key risk: pruning. Make this risk visible in tests and in code comments.

**Common pitfall:** Mutating shared state without undoing it correctly causes ghosts from one branch to corrupt another.

## Reference implementation sketch

```python
def permutations(items):
    items = list(items)
    used = [False] * len(items)
    out = []
    def search(path):
        if len(path) == len(items):
            out.append(tuple(path)); return
        for i, x in enumerate(items):
            if not used[i]:
                used[i] = True
                path.append(x)
                search(path)
                path.pop()
                used[i] = False
    search([])
    return out
```

# Chapter 25: String Algorithms and Automata

# Page 355 — Naive pattern matching

**Chapter 25: String Algorithms and Automata**  
**Angle:** proof idea

String algorithms avoid rechecking characters by preserving information about matched prefixes, suffixes, or automaton states.

## Core use
Use them for search, indexing, compression, plagiarism detection, DNA/protein sequence work, and text editors.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| often linear after preprocessing | preprocessing structures | encoding assumptions |

## Working notes
- A proof usually follows the implementation structure: initialization establishes the invariant, each step preserves it, and termination converts it into the postcondition.
- Key risk: encoding assumptions. Make this risk visible in tests and in code comments.

**Common pitfall:** Algorithms over characters can fail in user-facing text unless normalization and grapheme boundaries are considered.

## Reference implementation sketch

```python
def prefix_function(pattern):
    pi = [0] * len(pattern)
    j = 0
    for i in range(1, len(pattern)):
        while j and pattern[i] != pattern[j]:
            j = pi[j - 1]
        if pattern[i] == pattern[j]:
            j += 1
        pi[i] = j
    return pi

def kmp_search(text, pattern):
    if not pattern:
        return list(range(len(text) + 1))
    pi, j, hits = prefix_function(pattern), 0, []
    for i, ch in enumerate(text):
        while j and ch != pattern[j]:
            j = pi[j - 1]
        if ch == pattern[j]:
            j += 1
        if j == len(pattern):
            hits.append(i - j + 1)
            j = pi[j - 1]
    return hits
```

```python
def prefix_function(pattern):
    pi = [0] * len(pattern)
    j = 0
    for i in range(1, len(pattern)):
        while j and pattern[i] != pattern[j]:
            j = pi[j - 1]
        if pattern[i] == pattern[j]:
            j += 1
        pi[i] = j
    return pi

def kmp_search(text, pattern):
    if not pattern:
        return list(range(len(text) + 1))
    pi, j, hits = prefix_function(pattern), 0, []
    for i, ch in enumerate(text):
        while j and ch != pattern[j]:
            j = pi[j - 1]
        if ch == pattern[j]:
            j += 1
        if j == len(pattern):
            hits.append(i - j + 1)
            j = pi[j - 1]
    return hits
```

# Page 368 — Edit distance revisited

**Chapter 25: String Algorithms and Automata**  
**Angle:** applications

String algorithms avoid rechecking characters by preserving information about matched prefixes, suffixes, or automaton states.

## Core use
Use them for search, indexing, compression, plagiarism detection, DNA/protein sequence work, and text editors.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| states × transition cost | state table | state definition |

## Working notes
- Look for Edit distance revisited whenever data has structure: order, hierarchy, graph connectivity, repeated subproblems, locality, or constraints that rule out most candidates.
- Key risk: state definition. Make this risk visible in tests and in code comments.

**Common pitfall:** Algorithms over characters can fail in user-facing text unless normalization and grapheme boundaries are considered.

## Reference implementation sketch

```python
def lcs(a, b):
    n, m = len(a), len(b)
    dp = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(n - 1, -1, -1):
        for j in range(m - 1, -1, -1):
            if a[i] == b[j]:
                dp[i][j] = 1 + dp[i + 1][j + 1]
            else:
                dp[i][j] = max(dp[i + 1][j], dp[i][j + 1])
    return dp[0][0]
```

**Question:** What single assumption makes the usual Edit distance revisited analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time states × transition cost and space state table, but the exact assumption is the one stated in the page notes.

# Page 380 — Sieve of Eratosthenes

**Chapter 26: Number Theory and Combinatorics**  
**Angle:** mental model

Number-theoretic algorithms exploit divisibility, modular structure, and recurrence in integer objects. Combinatorics counts structured possibilities without enumerating them.

## Core use
Use these tools for cryptography, hashing, randomized tests, counting, modular DP, and competitive programming constraints.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| O(n log log n) | O(n) | limit memory |

## Working notes
- Think of Sieve of Eratosthenes as a number theory pattern with an explicit state, a transition rule, and a measure of progress.
- Key risk: limit memory. Make this risk visible in tests and in code comments.

**Common pitfall:** Modulo arithmetic changes division; you need inverses under the right conditions, not ordinary division.

## Reference implementation sketch

```python
def sieve(limit):
    is_prime = [True] * (limit + 1)
    is_prime[:2] = [False, False]
    p = 2
    while p * p <= limit:
        if is_prime[p]:
            for q in range(p * p, limit + 1, p):
                is_prime[q] = False
        p += 1
    return [i for i, ok in enumerate(is_prime) if ok]
```

**Answer:** The representation and invariant must match the intended model; the common cost summary is time O(n log log n) and space O(n), but the exact assumption is the one stated in the page notes.

# Page 385 — Combinations n choose k

**Chapter 26: Number Theory and Combinatorics**  
**Angle:** proof idea

Number-theoretic algorithms exploit divisibility, modular structure, and recurrence in integer objects. Combinatorics counts structured possibilities without enumerating them.

## Core use
Use these tools for cryptography, hashing, randomized tests, counting, modular DP, and competitive programming constraints.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| often logarithmic or sieve-like | varies | overflow and modulus mistakes |

## Working notes
- A proof usually follows the implementation structure: initialization establishes the invariant, each step preserves it, and termination converts it into the postcondition.
- Key risk: overflow and modulus mistakes. Make this risk visible in tests and in code comments.

**Common pitfall:** Modulo arithmetic changes division; you need inverses under the right conditions, not ordinary division.

## Reference implementation sketch

```python
def permutations(items):
    items = list(items)
    used = [False] * len(items)
    out = []
    def search(path):
        if len(path) == len(items):
            out.append(tuple(path)); return
        for i, x in enumerate(items):
            if not used[i]:
                used[i] = True
                path.append(x)
                search(path)
                path.pop()
                used[i] = False
    search([])
    return out
```

**Question:** What single assumption makes the usual Combinations n choose k analysis valid?

# Page 395 — Points, vectors, and orientation

**Chapter 27: Computational Geometry**  
**Angle:** proof idea

Geometry algorithms turn spatial relationships into predicates such as orientation, intersection, containment, and distance.

## Core use
Use them for mapping, graphics, robotics, collision detection, spatial databases, and layout problems.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| often O(n log n) | O(n) | precision errors |

## Working notes
- A proof usually follows the implementation structure: initialization establishes the invariant, each step preserves it, and termination converts it into the postcondition.
- Key risk: precision errors. Make this risk visible in tests and in code comments.

**Common pitfall:** Floating-point comparisons near zero can flip orientation tests and destroy an otherwise correct algorithm.

## Reference implementation sketch

```python
def orientation(a, b, c):
    # positive: counterclockwise, negative: clockwise, zero: collinear
    return (b[0] - a[0]) * (c[1] - a[1]) - (b[1] - a[1]) * (c[0] - a[0])

def convex_hull(points):
    pts = sorted(set(points))
    if len(pts) <= 1:
        return pts
    lower = []
    for p in pts:
        while len(lower) >= 2 and orientation(lower[-2], lower[-1], p) <= 0:
            lower.pop()
        lower.append(p)
    upper = []
    for p in reversed(pts):
        while len(upper) >= 2 and orientation(upper[-2], upper[-1], p) <= 0:
            upper.pop()
        upper.append(p)
    return lower[:-1] + upper[:-1]
```

## Checkpoint quiz
**Question:** What single assumption makes the usual Points, vectors, and orientation analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time often O(n log n) and space O(n), but the exact assumption is the one stated in the page notes.

# Page 404 — Rectangle union area

**Chapter 27: Computational Geometry**  
**Angle:** edge cases

Geometry algorithms turn spatial relationships into predicates such as orientation, intersection, containment, and distance.

## Core use
Use them for mapping, graphics, robotics, collision detection, spatial databases, and layout problems.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| often O(n log n) | O(n) | precision errors |

## Working notes
- Test the empty input, one-element input, duplicates, already-solved input, reverse input, disconnected components, and maximum-size input where relevant.
- Key risk: precision errors. Make this risk visible in tests and in code comments.

**Common pitfall:** Floating-point comparisons near zero can flip orientation tests and destroy an otherwise correct algorithm.

## Reference implementation sketch

```python
class DSU:
    def __init__(self, n):
        self.parent = list(range(n))
        self.size = [1] * n
    def find(self, x):
        while x != self.parent[x]:
            self.parent[x] = self.parent[self.parent[x]]
            x = self.parent[x]
        return x
    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False
        if self.size[ra] < self.size[rb]:
            ra, rb = rb, ra
        self.parent[rb] = ra
        self.size[ra] += self.size[rb]
        return True
```

**Question:** What single assumption makes the usual Rectangle union area analysis valid?

**Answer:** The representation and invariant must match the intended model; the common cost summary is time often O(n log n) and space O(n), but the exact assumption is the one stated in the page notes.

```python
def convex_hull(points):
    pts = sorted(set(points))
    if len(pts) <= 1:
        return pts
    lower = []
    for p in pts:
        while len(lower) >= 2 and orientation(lower[-2], lower[-1], p) <= 0:
            lower.pop()
        lower.append(p)
    upper = []
    for p in reversed(pts):
        while len(upper) >= 2 and orientation(upper[-2], upper[-1], p) <= 0:
            upper.pop()
        upper.append(p)
    return lower[:-1] + upper[:-1]
```

```python
import random

def reservoir_sample(stream, k):
    sample = []
    for i, item in enumerate(stream, start=1):
        if i <= k:
            sample.append(item)
        else:
            j = random.randrange(i)
            if j < k:
                sample[j] = item
    return sample
```

# Page 413 — Randomized quicksort

**Chapter 28: Randomized, Approximation, and Streaming**  
**Angle:** complexity

These algorithms trade determinism, exactness, or full memory access for speed, simplicity, scalability, or robustness.

## Core use
Use them when exact solutions are too expensive, inputs are massive, or probability gives strong guarantees with small memory.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| goal-dependent | often sublinear memory | probability assumptions |

## Working notes
- Budget the cost as goal-dependent time and often sublinear memory space in the common model, then revisit the bound when the input representation changes.
- Key risk: probability assumptions. Make this risk visible in tests and in code comments.

**Common pitfall:** A randomized or approximate algorithm still needs a contract: success probability, error bound, and input assumptions.

## Reference implementation sketch

```python
def quicksort(a):
    if len(a) <= 1:
        return a[:]
    pivot = a[len(a) // 2]
    less  = [x for x in a if x < pivot]
    equal = [x for x in a if x == pivot]
    more  = [x for x in a if x > pivot]
    return quicksort(less) + equal + quicksort(more)
```

```python
    return chosen
```

# Page 473 — A* in AI planning

**Chapter 30: Systems, Databases, Compression, Crypto, and AI Algorithms**  
**Angle:** complexity

Systems algorithms make resource tradeoffs explicit: time, space, latency, bandwidth, reliability, and security constraints shape the design.

## Core use
Use these topics to connect classical algorithms with storage engines, operating systems, compression formats, security protocols, and intelligent search.

## Complexity lens

| Time | Space | Main risk |
|---|---|---|
| domain-dependent | domain-dependent | ignoring operational constraints |

## Working notes
- Budget the cost as domain-dependent time and domain-dependent space in the common model, then revisit the bound when the input representation changes.
- Key risk: ignoring operational constraints. Make this risk visible in tests and in code comments.

**Common pitfall:** A theoretically efficient algorithm may be unusable if it leaks secrets, thrashes cache, or violates consistency assumptions.

## Reference implementation sketch

```python
import heapq

def dijkstra(graph, source):
    dist = {source: 0}
    parent = {source: None}
    pq = [(0, source)]
    while pq:
        d, u = heapq.heappop(pq)
        if d != dist.get(u):
            continue
        for v, w in graph[u]:
            nd = d + w
            if nd < dist.get(v, float('inf')):
                dist[v] = nd
                parent[v] = u
                heapq.heappush(pq, (nd, v))
    return dist, parent
```

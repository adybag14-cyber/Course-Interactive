## Chapter 01 deep dive — Orientation and Algorithmic Thinking

**Anchor page:** 001  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A reusable workflow object that stores the problem statement, invariant, edge cases, and validation checks before any optimization happens.

**Useful links:** MIT 6.006 overview (https://ocw.mit.edu/courses/6-006-introduction-to-algorithms-spring-2020/), CP-Algorithms index (https://cp-algorithms.com/), Design recipe mindset (https://htdp.org/2023-8-14/Book/index.html)



````python
from dataclasses import dataclass, field
from typing import Any, Callable

@dataclass
class AlgorithmContract:
    problem: str
    representation: str
    invariant: str
    edge_cases: list[str] = field(default_factory=list)
    check: Callable[[Any], bool] | None = None

    def validate(self, candidate: Any) -> bool:
        if self.check is None:
            return True
        return self.check(candidate)

def solve_with_contract(data: list[int], contract: AlgorithmContract) -> dict:
    state = {
        'input': list(data),
        'representation': contract.representation,
        'invariant': contract.invariant,
        'edge_cases': list(contract.edge_cases),
    }
    state['empty'] = len(data) == 0
    state['one_item'] = len(data) == 1
    state['sorted_preview'] = sorted(data)
    state['valid'] = contract.validate(state['sorted_preview'])
    return state

contract = AlgorithmContract(
    problem='Sort a list of integers',
    representation='Python list of comparable integers',
    invariant='sorted_preview is a permutation of input',
    edge_cases=['empty input', 'single item', 'duplicates'],
    check=lambda out: out == sorted(out),
)

print(solve_with_contract([5, 1, 5, 3], contract))

````

## Chapter 02 deep dive — Math and Complexity Foundations

**Anchor page:** 011  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** Compare two functions over increasing input sizes and report which one dominates empirically. This turns asymptotic language into something you can inspect.

**Useful links:** Big-O primer (https://cp-algorithms.com/algebra/big-integer.html), Python math module (https://docs.python.org/3/library/math.html), MIT mathematics for CS (https://ocw.mit.edu/courses/6-042j-mathematics-for-computer-science-fall-2010/)



````python
import math
from typing import Callable

def compare_growth(f: Callable[[int], float], g: Callable[[int], float], sizes: list[int]) -> list[tuple[int, float, float, str]]:
    rows = []
    for n in sizes:
        left = f(n)
        right = g(n)
        winner = 'f' if left < right else 'g' if right < left else 'tie'
        rows.append((n, left, right, winner))
    return rows

f = lambda n: n * math.log2(max(2, n))
g = lambda n: n * n
for row in compare_growth(f, g, [2, 4, 8, 16, 32, 64]):
    print(row)

````

## Chapter 03 deep dive — Correctness, Proofs, and Testing

**Anchor page:** 029  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A concrete proof harness: the implementation exposes the loop invariant and a brute-force checker so the proof and the code stay connected.

**Useful links:** pytest docs (https://docs.pytest.org/en/stable/), Hypothesis docs (https://hypothesis.readthedocs.io/en/latest/), CP-Algorithms correctness style (https://cp-algorithms.com/)



````python
def insertion_sort(values: list[int]) -> list[int]:
    arr = list(values)
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
        assert arr[: i + 1] == sorted(arr[: i + 1])
    return arr

def brute_force_sorted(values: list[int]) -> list[int]:
    return sorted(values)

samples = [[], [1], [3, 2, 1], [5, 1, 5, 3]]
for sample in samples:
    got = insertion_sort(sample)
    want = brute_force_sorted(sample)
    assert got == want
    print(sample, '->', got)

````

## Chapter 04 deep dive — Python and JavaScript Algorithm Tooling

**Anchor page:** 041  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A small benchmark harness for algorithms. It records time, validates output, and gives one place to compare implementations fairly.

**Useful links:** timeit docs (https://docs.python.org/3/library/timeit.html), itertools docs (https://docs.python.org/3/library/itertools.html), MDN JavaScript guide (https://developer.mozilla.org/en-US/docs/Web/JavaScript/Guide)



````python
import random
import time
from typing import Callable

def benchmark(fn: Callable[[list[int]], list[int]], size: int, repeats: int = 3) -> dict:
    timings = []
    for _ in range(repeats):
        data = [random.randint(0, 10_000) for _ in range(size)]
        start = time.perf_counter()
        out = fn(data)
        timings.append(time.perf_counter() - start)
        assert out == sorted(data)
    return {'size': size, 'best_seconds': min(timings), 'runs': repeats}

print(benchmark(lambda xs: sorted(xs), 5000))

````

## Chapter 05 deep dive — Arrays, Strings, and Memory

**Anchor page:** 051  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A foundational array technique: preprocess once, then answer many interval queries in O(1) time.

**Useful links:** Python bisect docs (https://docs.python.org/3/library/bisect.html), Two pointers and scans (https://cp-algorithms.com/others/tortoise_and_hare.html), Unicode HOWTO (https://docs.python.org/3/howto/unicode.html)



````python
class PrefixSums:
    def __init__(self, values: list[int]) -> None:
        self.pref = [0]
        for value in values:
            self.pref.append(self.pref[-1] + value)

    def range_sum(self, left: int, right: int) -> int:
        return self.pref[right + 1] - self.pref[left]

ps = PrefixSums([3, 1, 4, 1, 5, 9])
print(ps.range_sum(1, 3))  # 1 + 4 + 1
print(ps.range_sum(0, 5))

````

## Chapter 06 deep dive — Search and Binary Search

**Anchor page:** 067  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** The canonical template for finding the first position whose value is at least a target.

**Useful links:** Python bisect (https://docs.python.org/3/library/bisect.html), CP-Algorithms binary search (https://cp-algorithms.com/num_methods/binary_search.html), MIT 6.006 notes (https://ocw.mit.edu/courses/6-006-introduction-to-algorithms-spring-2020/)



````python
def lower_bound(arr: list[int], target: int) -> int:
    lo, hi = 0, len(arr)
    while lo < hi:
        mid = (lo + hi) // 2
        if arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid
    return lo

data = [2, 4, 4, 7, 9, 12]
for t in [0, 4, 5, 12, 20]:
    idx = lower_bound(data, t)
    print(t, idx)

````

## Chapter 07 deep dive — Sorting Algorithms

**Anchor page:** 079  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A full divide-and-conquer sorter that emphasizes stable merging and a clean recurrence.

**Useful links:** Python sorting HOWTO (https://docs.python.org/3/howto/sorting.html), CP-Algorithms sorting overview (https://cp-algorithms.com/), Timsort note (https://en.wikipedia.org/wiki/Timsort)



````python
def merge(left: list[int], right: list[int]) -> list[int]:
    out = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            out.append(left[i])
            i += 1
        else:
            out.append(right[j])
            j += 1
    out.extend(left[i:])
    out.extend(right[j:])
    return out

def merge_sort(arr: list[int]) -> list[int]:
    if len(arr) <= 1:
        return list(arr)
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

print(merge_sort([31, 2, 13, 24, 35, 6, 17]))

````

## Chapter 08 deep dive — Recursion and Divide and Conquer

**Anchor page:** 097  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A classic recurrence where the answer is either in the left half, the right half, or crossing the midpoint.

**Useful links:** Recursion HOWTO (https://docs.python.org/3/tutorial/controlflow.html#defining-functions), CP-Algorithms divide and conquer (https://cp-algorithms.com/), Master theorem overview (https://en.wikipedia.org/wiki/Master_theorem_(analysis_of_algorithms))



````python
def max_crossing(arr: list[int], lo: int, mid: int, hi: int) -> int:
    best_left = cur = float('-inf')
    s = 0
    for i in range(mid, lo - 1, -1):
        s += arr[i]
        best_left = max(best_left, s)
    best_right = cur = float('-inf')
    s = 0
    for i in range(mid + 1, hi + 1):
        s += arr[i]
        best_right = max(best_right, s)
    return best_left + best_right

def max_subarray(arr: list[int], lo: int = 0, hi: int | None = None) -> int:
    if hi is None:
        hi = len(arr) - 1
    if lo == hi:
        return arr[lo]
    mid = (lo + hi) // 2
    return max(
        max_subarray(arr, lo, mid),
        max_subarray(arr, mid + 1, hi),
        max_crossing(arr, lo, mid, hi),
    )

print(max_subarray([2, -1, 3, -5, 4, 6, -2]))

````

## Chapter 09 deep dive — Stacks, Queues, Deques, and Linked Lists

**Anchor page:** 111  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A double-ended queue implemented with a doubly linked list so pushes and pops are O(1) at both ends.

**Useful links:** collections.deque docs (https://docs.python.org/3/library/collections.html#collections.deque), Linked list overview (https://en.wikipedia.org/wiki/Linked_list), CP-Algorithms data structures (https://cp-algorithms.com/data_structures/)



````python
class Node:
    def __init__(self, value: int):
        self.value = value
        self.prev = None
        self.next = None

class LinkedDeque:
    def __init__(self):
        self.head = self.tail = None

    def append(self, value: int) -> None:
        node = Node(value)
        if not self.tail:
            self.head = self.tail = node
            return
        node.prev = self.tail
        self.tail.next = node
        self.tail = node

    def appendleft(self, value: int) -> None:
        node = Node(value)
        if not self.head:
            self.head = self.tail = node
            return
        node.next = self.head
        self.head.prev = node
        self.head = node

    def popleft(self) -> int:
        if not self.head:
            raise IndexError('empty deque')
        value = self.head.value
        self.head = self.head.next
        if self.head:
            self.head.prev = None
        else:
            self.tail = None
        return value

dq = LinkedDeque()
dq.append(2); dq.append(3); dq.appendleft(1)
print(dq.popleft(), dq.popleft())

````

## Chapter 10 deep dive — Hashing and Hash Tables

**Anchor page:** 127  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A bare-bones hash table that makes collisions visible and helps explain load factor and resizing.

**Useful links:** Python dict docs (https://docs.python.org/3/tutorial/datastructures.html#dictionaries), CP-Algorithms string hashing (https://cp-algorithms.com/string/string-hashing.html), Hash table overview (https://en.wikipedia.org/wiki/Hash_table)



````python
class LinearProbingMap:
    def __init__(self, capacity: int = 8):
        self.slots = [None] * capacity
        self.size = 0

    def _find_slot(self, key: str) -> int:
        idx = hash(key) % len(self.slots)
        while self.slots[idx] is not None and self.slots[idx][0] != key:
            idx = (idx + 1) % len(self.slots)
        return idx

    def put(self, key: str, value: int) -> None:
        idx = self._find_slot(key)
        if self.slots[idx] is None:
            self.size += 1
        self.slots[idx] = (key, value)

    def get(self, key: str) -> int:
        idx = self._find_slot(key)
        if self.slots[idx] is None:
            raise KeyError(key)
        return self.slots[idx][1]

table = LinearProbingMap()
table.put('red', 3)
table.put('blue', 5)
print(table.get('red'))

````

## Chapter 11 deep dive — Trees and Traversals

**Anchor page:** 141  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** Preorder, inorder, and postorder traversal in a way that makes the call stack explicit.

**Useful links:** Python heapq docs (https://docs.python.org/3/library/heapq.html), Tree traversal overview (https://en.wikipedia.org/wiki/Tree_traversal), CP-Algorithms DFS (https://cp-algorithms.com/graph/depth-first-search.html)



````python
class Node:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right

def preorder(root):
    if not root:
        return []
    out, stack = [], [root]
    while stack:
        node = stack.pop()
        out.append(node.value)
        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)
    return out

tree = Node(2, Node(1), Node(4, Node(3), Node(5)))
print(preorder(tree))

````

## Chapter 12 deep dive — Priority Queues and Heaps

**Anchor page:** 159  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A textbook heap with sift-up and sift-down. It exposes the array representation behind priority queues.

**Useful links:** heapq docs (https://docs.python.org/3/library/heapq.html), Priority queue overview (https://en.wikipedia.org/wiki/Priority_queue), CP-Algorithms dijkstra (https://cp-algorithms.com/graph/dijkstra.html)



````python
class MinHeap:
    def __init__(self):
        self.data = []

    def push(self, value: int) -> None:
        self.data.append(value)
        i = len(self.data) - 1
        while i > 0:
            p = (i - 1) // 2
            if self.data[p] <= self.data[i]:
                break
            self.data[p], self.data[i] = self.data[i], self.data[p]
            i = p

    def pop(self) -> int:
        if not self.data:
            raise IndexError('empty heap')
        self.data[0], self.data[-1] = self.data[-1], self.data[0]
        value = self.data.pop()
        i = 0
        while True:
            left = 2 * i + 1
            right = left + 1
            smallest = i
            if left < len(self.data) and self.data[left] < self.data[smallest]:
                smallest = left
            if right < len(self.data) and self.data[right] < self.data[smallest]:
                smallest = right
            if smallest == i:
                break
            self.data[i], self.data[smallest] = self.data[smallest], self.data[i]
            i = smallest
        return value

heap = MinHeap()
for x in [7, 3, 9, 1, 4]:
    heap.push(x)
print([heap.pop() for _ in range(5)])

````

## Chapter 13 deep dive — Heaps and Priority Queues

**Anchor page:** 175  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** This implementation focuses on the BST invariant. It is intentionally unbalanced so the need for balancing stays visible.

**Useful links:** bisect docs (https://docs.python.org/3/library/bisect.html), BST overview (https://en.wikipedia.org/wiki/Binary_search_tree), CP-Algorithms segment tree intro (https://cp-algorithms.com/data_structures/segment_tree.html)



````python
class BSTNode:
    def __init__(self, key: int):
        self.key = key
        self.left = None
        self.right = None

    def insert(self, key: int) -> None:
        if key < self.key:
            if self.left:
                self.left.insert(key)
            else:
                self.left = BSTNode(key)
        elif key > self.key:
            if self.right:
                self.right.insert(key)
            else:
                self.right = BSTNode(key)

    def contains(self, key: int) -> bool:
        if key == self.key:
            return True
        if key < self.key:
            return self.left.contains(key) if self.left else False
        return self.right.contains(key) if self.right else False

root = BSTNode(8)
for x in [3, 10, 1, 6, 14, 4, 7, 13]:
    root.insert(x)
print(root.contains(7), root.contains(5))

````

## Chapter 14 deep dive — Disjoint Set Union

**Anchor page:** 189  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A standard disjoint-set union structure used for connectivity and Kruskal-style algorithms.

**Useful links:** CP-Algorithms DSU (https://cp-algorithms.com/data_structures/disjoint_set_union.html), Union-find overview (https://en.wikipedia.org/wiki/Disjoint-set_data_structure), Python list docs (https://docs.python.org/3/tutorial/datastructures.html)



````python
class DSU:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.size = [1] * n

    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, a: int, b: int) -> bool:
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False
        if self.size[ra] < self.size[rb]:
            ra, rb = rb, ra
        self.parent[rb] = ra
        self.size[ra] += self.size[rb]
        return True

dsu = DSU(6)
for edge in [(0, 1), (1, 2), (3, 4), (2, 4)]:
    dsu.union(*edge)
print([dsu.find(i) for i in range(6)])

````

````python
# Page 198: DSU pitfalls
EXAMPLE = {'page': 198, 'topic': 'DSU pitfalls', 'chapter': 'Disjoint Set Union', 'kind': 'dsu', 'scenario': 'Run a concrete dsu miniature for DSU pitfalls.', 'sample_input': {'values': [27, 38, 9, 20, 31, 2, 13], 'items': ['red', 'blue', 'red', 'green', 'blue', 'red'], 'edges': [(0, 1), (1, 2), (3, 4), (2, 4)], 'tree': {'A': ['B', 'C'], 'B': ['D', 'E'], 'C': ['F']}, 'requests': ['A', 'B', 'A', 'C', 'D', 'A']}, 'expected_output': 'Inspectable result for DSU pitfalls', 'watch_for': 'The invariant that makes DSU pitfalls safe is checked before the next transition.', 'try_next': 'Change the sample input and predict which line in the dsu example changes first.', 'docs': []}
SECOND_EXAMPLE = dict(EXAMPLE, sample_input=dict(EXAMPLE['sample_input'], page_marker='198-dsu_pitfalls-variant'))
EDGE_CASES = ['empty input', 'single item', 'duplicate-heavy input', 'boundary-size input']


class DSU:
    def __init__(self, n): self.parent=list(range(n)); self.size=[1]*n
    def find(self, x):
        while self.parent[x] != x:
            self.parent[x] = self.parent[self.parent[x]]; x = self.parent[x]
        return x
    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb: return False
        if self.size[ra] < self.size[rb]: ra, rb = rb, ra
        self.parent[rb] = ra; self.size[ra] += self.size[rb]; return True

def dsu_pitfalls_p198(n, edges):
    dsu = DSU(n); accepted=[]
    for edge in edges:
        if dsu.union(*edge): accepted.append(edge)
    return accepted, [dsu.find(i) for i in range(n)]

def run_example(example=EXAMPLE):
    return dsu_pitfalls_p198(5, example['sample_input']['edges'])

def run_edge_cases():
    return {'topic': EXAMPLE['topic'], 'edge_cases': EDGE_CASES, 'second_marker': SECOND_EXAMPLE['sample_input'].get('page_marker')}

run_example()
````

## Chapter 15 deep dive — Graph Basics and Traversal

**Anchor page:** 199  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** Breadth-first search simultaneously teaches traversal order, queue discipline, and shortest path layers.

**Useful links:** collections.deque docs (https://docs.python.org/3/library/collections.html#collections.deque), CP-Algorithms BFS (https://cp-algorithms.com/graph/breadth-first-search.html), NetworkX docs (https://networkx.org/documentation/stable/)



````python
from collections import deque

def bfs_distances(graph: dict[str, list[str]], start: str) -> dict[str, int]:
    dist = {start: 0}
    q = deque([start])
    while q:
        node = q.popleft()
        for nxt in graph.get(node, []):
            if nxt not in dist:
                dist[nxt] = dist[node] + 1
                q.append(nxt)
    return dist

graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['D', 'E'],
    'D': ['F'],
    'E': [],
    'F': [],
}
print(bfs_distances(graph, 'A'))

````

## Chapter 16 deep dive — DAGs and Topological Algorithms

**Anchor page:** 217  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A deterministic way to order DAG vertices by repeatedly removing indegree-zero nodes.

**Useful links:** CP-Algorithms topological sort (https://cp-algorithms.com/graph/topological-sort.html), collections.deque (https://docs.python.org/3/library/collections.html#collections.deque), DAG overview (https://en.wikipedia.org/wiki/Directed_acyclic_graph)



````python
from collections import deque

def topological_sort(graph: dict[str, list[str]]) -> list[str]:
    indeg = {u: 0 for u in graph}
    for u in graph:
        for v in graph[u]:
            indeg[v] = indeg.get(v, 0) + 1
    q = deque([u for u, d in indeg.items() if d == 0])
    order = []
    while q:
        u = q.popleft()
        order.append(u)
        for v in graph.get(u, []):
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)
    if len(order) != len(indeg):
        raise ValueError('graph contains a cycle')
    return order

dag = {'shirt': ['belt', 'tie'], 'tie': ['jacket'], 'belt': ['jacket'], 'jacket': [], 'watch': []}
print(topological_sort(dag))

````

## Chapter 17 deep dive — Shortest Paths

**Anchor page:** 227  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A standard shortest-path algorithm for nonnegative edge weights, using a priority queue and distance relaxations.

**Useful links:** CP-Algorithms Dijkstra (https://cp-algorithms.com/graph/dijkstra.html), heapq docs (https://docs.python.org/3/library/heapq.html), Shortest path overview (https://en.wikipedia.org/wiki/Shortest_path_problem)



````python
import heapq

def dijkstra(graph: dict[str, list[tuple[str, int]]], start: str) -> dict[str, int]:
    dist = {start: 0}
    pq = [(0, start)]
    while pq:
        d, u = heapq.heappop(pq)
        if d != dist[u]:
            continue
        for v, w in graph.get(u, []):
            nd = d + w
            if nd < dist.get(v, float('inf')):
                dist[v] = nd
                heapq.heappush(pq, (nd, v))
    return dist

graph = {
    'A': [('B', 4), ('C', 1)],
    'B': [('D', 1)],
    'C': [('B', 2), ('D', 5)],
    'D': []
}
print(dijkstra(graph, 'A'))

````

## Chapter 18 deep dive — Spanning Trees and Connectivity

**Anchor page:** 245  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** The MST emerges by scanning edges in increasing order and skipping any edge that would create a cycle.

**Useful links:** CP-Algorithms Kruskal (https://cp-algorithms.com/graph/mst_kruskal.html), Priority queues and heaps (https://docs.python.org/3/library/heapq.html), MST overview (https://en.wikipedia.org/wiki/Minimum_spanning_tree)



````python
class DSU:
    def __init__(self, n: int):
        self.parent = list(range(n))
    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    def union(self, a: int, b: int) -> bool:
        a, b = self.find(a), self.find(b)
        if a == b:
            return False
        self.parent[b] = a
        return True

def kruskal(n: int, edges: list[tuple[int, int, int]]) -> tuple[int, list[tuple[int, int, int]]]:
    dsu = DSU(n)
    total = 0
    chosen = []
    for w, u, v in sorted(edges):
        if dsu.union(u, v):
            total += w
            chosen.append((u, v, w))
    return total, chosen

edges = [(1, 0, 1), (4, 0, 2), (2, 1, 2), (7, 1, 3), (3, 2, 3)]
print(kruskal(4, edges))

````

## Chapter 19 deep dive — Network Flow and Matching

**Anchor page:** 257  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** Residual edges, augmenting paths, and bottleneck capacities are visible in this full maximum-flow implementation.

**Useful links:** CP-Algorithms Edmonds-Karp (https://cp-algorithms.com/graph/edmonds_karp.html), Flow overview (https://en.wikipedia.org/wiki/Maximum_flow_problem), collections.deque docs (https://docs.python.org/3/library/collections.html#collections.deque)



````python
from collections import deque, defaultdict

def edmonds_karp(capacity: dict[str, dict[str, int]], source: str, sink: str) -> int:
    residual = defaultdict(dict)
    for u, nbrs in capacity.items():
        for v, cap in nbrs.items():
            residual[u][v] = cap
            residual[v].setdefault(u, 0)
    flow = 0
    while True:
        parent = {source: None}
        q = deque([source])
        while q and sink not in parent:
            u = q.popleft()
            for v, cap in residual[u].items():
                if cap > 0 and v not in parent:
                    parent[v] = u
                    q.append(v)
        if sink not in parent:
            return flow
        aug = float('inf')
        v = sink
        while v != source:
            u = parent[v]
            aug = min(aug, residual[u][v])
            v = u
        v = sink
        while v != source:
            u = parent[v]
            residual[u][v] -= aug
            residual[v][u] += aug
            v = u
        flow += aug

capacity = {'s': {'a': 3, 'b': 2}, 'a': {'b': 1, 't': 2}, 'b': {'t': 3}, 't': {}}
print(edmonds_karp(capacity, 's', 't'))

````

## Chapter 20 deep dive — Dynamic Programming Foundations

**Anchor page:** 271  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A foundational dynamic program where each cell stores the best value obtainable from the first i items with capacity cap.

**Useful links:** CP-Algorithms knapsack (https://cp-algorithms.com/dynamic_programming/knapsack.html), DP overview (https://en.wikipedia.org/wiki/Dynamic_programming), functools docs (https://docs.python.org/3/library/functools.html)



````python
def interval_schedule(intervals: list[tuple[int, int]]) -> list[tuple[int, int]]:
    chosen = []
    end_time = float('-inf')
    for start, finish in sorted(intervals, key=lambda x: x[1]):
        if start >= end_time:
            chosen.append((start, finish))
            end_time = finish
    return chosen

meetings = [(1, 4), (3, 5), (0, 6), (5, 7), (8, 9), (5, 9)]
print(interval_schedule(meetings))

````

## Chapter 21 deep dive — Sequence and String Dynamic Programming

**Anchor page:** 289  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A classic sequence/string DP. Each cell stores the LCS length for prefixes of the two strings.

**Useful links:** DP overview (https://en.wikipedia.org/wiki/Dynamic_programming), difflib docs (https://docs.python.org/3/library/difflib.html), CP-Algorithms intro (https://cp-algorithms.com/)



````python
def lcs_length(a: str, b: str) -> int:
    dp = [[0] * (len(b) + 1) for _ in range(len(a) + 1)]
    for i in range(1, len(a) + 1):
        for j in range(1, len(b) + 1):
            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])
    return dp[-1][-1]

print(lcs_length('algorithm', 'logarithm'))

````

## Chapter 22 deep dive — Grid, Tree, and Bitmask Dynamic Programming

**Anchor page:** 307  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A representative bitmask DP: state = (visited subset, current node), transition = try the next unvisited node.

**Useful links:** Bitmask DP overview (https://cp-algorithms.com/), itertools docs (https://docs.python.org/3/library/itertools.html), Dynamic programming overview (https://en.wikipedia.org/wiki/Dynamic_programming)



````python
from functools import lru_cache

def tsp(dist: list[list[int]]) -> int:
    n = len(dist)

    @lru_cache(maxsize=None)
    def dp(mask: int, u: int) -> int:
        if mask == (1 << n) - 1:
            return dist[u][0]
        ans = float('inf')
        for v in range(n):
            if not (mask & (1 << v)):
                ans = min(ans, dist[u][v] + dp(mask | (1 << v), v))
        return ans

    return dp(1, 0)

dist = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0],
]
print(tsp(dist))

````

## Chapter 23 deep dive — Greedy Algorithms

**Anchor page:** 325  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A table-based algorithm for comparing two sequences and computing insert/delete/replace cost.

**Useful links:** difflib docs (https://docs.python.org/3/library/difflib.html), Context-free grammar overview (https://en.wikipedia.org/wiki/Context-free_grammar), CP-Algorithms edit distance (https://cp-algorithms.com/)



````python
def edit_distance(a: str, b: str) -> int:
    dp = [[0] * (len(b) + 1) for _ in range(len(a) + 1)]
    for i in range(len(a) + 1):
        dp[i][0] = i
    for j in range(len(b) + 1):
        dp[0][j] = j
    for i in range(1, len(a) + 1):
        for j in range(1, len(b) + 1):
            cost = 0 if a[i - 1] == b[j - 1] else 1
            dp[i][j] = min(
                dp[i - 1][j] + 1,
                dp[i][j - 1] + 1,
                dp[i - 1][j - 1] + cost,
            )
    return dp[-1][-1]

print(edit_distance('kitten', 'sitting'))

````

````python
# Page 329: Huffman coding
EXAMPLE = {'page': 329, 'topic': 'Huffman coding', 'chapter': 'Greedy Algorithms', 'kind': 'huffman', 'scenario': 'Run a concrete huffman miniature for Huffman coding.', 'sample_input': {'intervals': [(0, 3), (1, 2), (3, 5), (4, 7), (5, 9)], 'weights': {'a': 5, 'b': 9, 'c': 12, 'd': 13, 'e': 16, 'f': 45}, 'jobs': [('A', 2, 100), ('B', 1, 19), ('C', 2, 27), ('D', 1, 25)]}, 'expected_output': 'Inspectable result for Huffman coding', 'watch_for': 'The invariant that makes Huffman coding safe is checked before the next transition.', 'try_next': 'Change the sample input and predict which line in the huffman example changes first.', 'docs': []}
SECOND_EXAMPLE = dict(EXAMPLE, sample_input=dict(EXAMPLE['sample_input'], page_marker='329-huffman_coding-variant'))
EDGE_CASES = ['empty input', 'single item', 'duplicate-heavy input', 'boundary-size input']


import heapq

def huffman_coding_p329(weights):
    heap = [(w, symbol) for symbol, w in weights.items()]
    heapq.heapify(heap)
    while len(heap) > 1:
        w1, a = heapq.heappop(heap); w2, b = heapq.heappop(heap)
        heapq.heappush(heap, (w1+w2, (a,b)))
    return heap[0][1]

def run_example(example=EXAMPLE):
    return huffman_coding_p329(example['sample_input']['weights'])

def run_edge_cases():
    return {'topic': EXAMPLE['topic'], 'edge_cases': EDGE_CASES, 'second_marker': SECOND_EXAMPLE['sample_input'].get('page_marker')}

run_example()
````

## Chapter 24 deep dive — Backtracking and Combinatorial Search

**Anchor page:** 341  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** The search tree is explicit: choose a column, test constraints, recurse, then undo.

**Useful links:** itertools docs (https://docs.python.org/3/library/itertools.html), Backtracking overview (https://en.wikipedia.org/wiki/Backtracking), CP-Algorithms generating combinations (https://cp-algorithms.com/combinatorics/generating_combinations.html)



````python
def solve_n_queens(n: int) -> list[list[str]]:
    cols, diag1, diag2 = set(), set(), set()
    board = [['.'] * n for _ in range(n)]
    out = []

    def dfs(r: int) -> None:
        if r == n:
            out.append([''.join(row) for row in board])
            return
        for c in range(n):
            if c in cols or r - c in diag1 or r + c in diag2:
                continue
            cols.add(c); diag1.add(r - c); diag2.add(r + c)
            board[r][c] = 'Q'
            dfs(r + 1)
            board[r][c] = '.'
            cols.remove(c); diag1.remove(r - c); diag2.remove(r + c)

    dfs(0)
    return out

print(solve_n_queens(4)[0])

````

## Chapter 25 deep dive — String Algorithms and Automata

**Anchor page:** 355  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** KMP is a string-algorithms anchor because it separates pattern preprocessing from streaming text scan.

**Useful links:** CP-Algorithms prefix function (https://cp-algorithms.com/string/prefix-function.html), Regex HOWTO (https://docs.python.org/3/howto/regex.html), String methods (https://docs.python.org/3/library/stdtypes.html#string-methods)



````python
from collections import deque

def neighbors(code: str):
    for i, ch in enumerate(code):
        digit = int(ch)
        for delta in (-1, 1):
            nd = (digit + delta) % 10
            yield code[:i] + str(nd) + code[i + 1:]

def open_lock(start: str, target: str, blocked: set[str]) -> int:
    q = deque([(start, 0)])
    seen = {start}
    while q:
        state, dist = q.popleft()
        if state == target:
            return dist
        for nxt in neighbors(state):
            if nxt not in seen and nxt not in blocked:
                seen.add(nxt)
                q.append((nxt, dist + 1))
    return -1

print(open_lock('0000', '0202', {'0101', '0102'}))

````

## Chapter 26 deep dive — Number Theory and Combinatorics

**Anchor page:** 375  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** Use the sieve to precompute smallest prime factors, then factor numbers efficiently.

**Useful links:** math docs (https://docs.python.org/3/library/math.html), CP-Algorithms sieve (https://cp-algorithms.com/algebra/sieve-of-eratosthenes.html), Combinatorics overview (https://docs.python.org/3/library/itertools.html)



````python
def smallest_prime_factors(n: int) -> list[int]:
    spf = list(range(n + 1))
    for i in range(2, int(n ** 0.5) + 1):
        if spf[i] == i:
            for j in range(i * i, n + 1, i):
                if spf[j] == j:
                    spf[j] = i
    return spf

def factorize(x: int, spf: list[int]) -> list[int]:
    out = []
    while x > 1:
        out.append(spf[x])
        x //= spf[x]
    return out

spf = smallest_prime_factors(50)
print(factorize(36, spf))

````

## Chapter 27 deep dive — Computational Geometry

**Anchor page:** 395  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** Orientation tests decide whether the hull should keep or remove a point.

**Useful links:** math.atan2 docs (https://docs.python.org/3/library/math.html#math.atan2), CP-Algorithms convex hull (https://cp-algorithms.com/geometry/convex-hull.html), Geometry overview (https://en.wikipedia.org/wiki/Convex_hull_algorithms)



````python
def cross(o, a, b):
    return (a[0] - o[0]) * (b[1] - o[1]) - (a[1] - o[1]) * (b[0] - o[0])

def convex_hull(points: list[tuple[int, int]]) -> list[tuple[int, int]]:
    pts = sorted(set(points))
    if len(pts) <= 1:
        return pts
    lower = []
    for p in pts:
        while len(lower) >= 2 and cross(lower[-2], lower[-1], p) <= 0:
            lower.pop()
        lower.append(p)
    upper = []
    for p in reversed(pts):
        while len(upper) >= 2 and cross(upper[-2], upper[-1], p) <= 0:
            upper.pop()
        upper.append(p)
    return lower[:-1] + upper[:-1]

print(convex_hull([(0, 0), (1, 1), (2, 0), (1, 2), (0, 2)]))

````

````python
# Page 404: Rectangle union area
EXAMPLE = {'page': 404, 'topic': 'Rectangle union area', 'chapter': 'Computational Geometry', 'kind': 'dsu', 'scenario': 'Run a concrete dsu miniature for Rectangle union area.', 'sample_input': {'values': [29, 40, 11, 22, 33, 4, 15], 'items': ['red', 'blue', 'red', 'green', 'blue', 'red'], 'edges': [(0, 1), (1, 2), (3, 4), (2, 4)], 'tree': {'A': ['B', 'C'], 'B': ['D', 'E'], 'C': ['F']}, 'requests': ['A', 'B', 'A', 'C', 'D', 'A']}, 'expected_output': 'Inspectable result for Rectangle union area', 'watch_for': 'The invariant that makes Rectangle union area safe is checked before the next transition.', 'try_next': 'Change the sample input and predict which line in the dsu example changes first.', 'docs': []}
SECOND_EXAMPLE = dict(EXAMPLE, sample_input=dict(EXAMPLE['sample_input'], page_marker='404-rectangle_union_ar-variant'))
EDGE_CASES = ['empty input', 'single item', 'duplicate-heavy input', 'boundary-size input']


class DSU:
    def __init__(self, n): self.parent=list(range(n)); self.size=[1]*n
    def find(self, x):
        while self.parent[x] != x:
            self.parent[x] = self.parent[self.parent[x]]; x = self.parent[x]
        return x
    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb: return False
        if self.size[ra] < self.size[rb]: ra, rb = rb, ra
        self.parent[rb] = ra; self.size[ra] += self.size[rb]; return True

def rectangle_union_area_p404(n, edges):
    dsu = DSU(n); accepted=[]
    for edge in edges:
        if dsu.union(*edge): accepted.append(edge)
    return accepted, [dsu.find(i) for i in range(n)]

def run_example(example=EXAMPLE):
    return rectangle_union_area_p404(5, example['sample_input']['edges'])

def run_edge_cases():
    return {'topic': EXAMPLE['topic'], 'edge_cases': EDGE_CASES, 'second_marker': SECOND_EXAMPLE['sample_input'].get('page_marker')}

run_example()
````

## Chapter 28 deep dive — Randomized, Approximation, and Streaming

**Anchor page:** 411  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** Choose a uniform random sample from a stream without storing the full stream.

**Useful links:** random docs (https://docs.python.org/3/library/random.html), Reservoir sampling overview (https://en.wikipedia.org/wiki/Reservoir_sampling), Streaming sketches intro (https://en.wikipedia.org/wiki/Streaming_algorithm)



````python
import random

def reservoir_sample(stream, k: int):
    sample = []
    for i, item in enumerate(stream):
        if i < k:
            sample.append(item)
        else:
            j = random.randint(0, i)
            if j < k:
                sample[j] = item
    return sample

random.seed(7)
print(reservoir_sample(range(1, 21), 5))

````

## Chapter 29 deep dive — Parallel, Distributed, and External-Memory Algorithms

**Anchor page:** 431  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A sequential simulation of the upsweep/downsweep idea so the parallel structure becomes visible before parallel execution.

**Useful links:** itertools.accumulate (https://docs.python.org/3/library/itertools.html#itertools.accumulate), MapReduce overview (https://en.wikipedia.org/wiki/MapReduce), B-tree overview (https://en.wikipedia.org/wiki/B-tree)



````python
def prefix_scan(values: list[int]) -> list[int]:
    out = [0] * len(values)
    running = 0
    for i, value in enumerate(values):
        running += value
        out[i] = running
    return out

blocks = [5, 1, 4, 2, 3]
print(prefix_scan(blocks))

````

## Chapter 30 deep dive — Systems, Databases, Compression, Crypto, and AI Algorithms

**Anchor page:** 447  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** One concrete system algorithm that ties together recency, hashing, linked order, and constant-time updates.

**Useful links:** functools.lru_cache (https://docs.python.org/3/library/functools.html#functools.lru_cache), OrderedDict docs (https://docs.python.org/3/library/collections.html#collections.OrderedDict), Huffman coding overview (https://en.wikipedia.org/wiki/Huffman_coding)



````python
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.data = OrderedDict()

    def get(self, key: str) -> int | None:
        if key not in self.data:
            return None
        self.data.move_to_end(key)
        return self.data[key]

    def put(self, key: str, value: int) -> None:
        if key in self.data:
            self.data.move_to_end(key)
        self.data[key] = value
        if len(self.data) > self.capacity:
            self.data.popitem(last=False)

cache = LRUCache(2)
cache.put('A', 1)
cache.put('B', 2)
cache.get('A')
cache.put('C', 3)
print(list(cache.data.items()))

````

````python
# Page 447: LRU and LFU caches
EXAMPLE = {'page': 447, 'topic': 'LRU and LFU caches', 'chapter': 'Systems, Databases, Compression, Crypto, and AI Algorithms', 'kind': 'lru_cache_impl', 'scenario': 'Run a concrete lru cache impl miniature for LRU and LFU caches.', 'sample_input': {'values': [10, 21, 32, 3, 14, 25, 36], 'items': ['red', 'blue', 'red', 'green', 'blue', 'red'], 'edges': [(0, 1), (1, 2), (3, 4), (2, 4)], 'tree': {'A': ['B', 'C'], 'B': ['D', 'E'], 'C': ['F']}, 'requests': ['A', 'B', 'A', 'C', 'D', 'A']}, 'expected_output': 'Inspectable result for LRU and LFU caches', 'watch_for': 'The invariant that makes LRU and LFU caches safe is checked before the next transition.', 'try_next': 'Change the sample input and predict which line in the lru cache impl example changes first.', 'docs': [('Python functools.cache/lru_cache', 'https://docs.python.org/3/library/functools.html')]}
SECOND_EXAMPLE = dict(EXAMPLE, sample_input=dict(EXAMPLE['sample_input'], page_marker='447-lru_and_lfu_caches-variant'))
EDGE_CASES = ['empty input', 'single item', 'duplicate-heavy input', 'boundary-size input']


from collections import OrderedDict

def lru_and_lfu_caches_p447(requests, capacity=3):
    cache, events = OrderedDict(), []
    for key in requests:
        hit = key in cache
        if hit: cache.move_to_end(key)
        else:
            if len(cache) == capacity: events.append(('evict', cache.popitem(last=False)[0]))
            cache[key] = True
        events.append(('hit' if hit else 'miss', key, list(cache)))
    return events

def run_example(example=EXAMPLE):
    return lru_and_lfu_caches_p447(example['sample_input']['requests'])

def run_edge_cases():
    return {'topic': EXAMPLE['topic'], 'edge_cases': EDGE_CASES, 'second_marker': SECOND_EXAMPLE['sample_input'].get('page_marker')}

run_example()
````

## Chapter 31 deep dive — Practice Projects and Reference

**Anchor page:** 487  
**Deep-dive asset:** full textbook implementation + custom chapter diagram.

**What this chapter implementation teaches:** A practical reference tool that tracks done pages, next review dates, and weak topics.

**Useful links:** datetime docs (https://docs.python.org/3/library/datetime.html), csv docs (https://docs.python.org/3/library/csv.html), collections.Counter (https://docs.python.org/3/library/collections.html#collections.Counter)



````python
from collections import Counter
from datetime import date, timedelta

def build_review_plan(done_pages: list[int], weak_topics: list[str]) -> dict:
    today = date.today()
    counts = Counter(weak_topics)
    return {
        'done_count': len(done_pages),
        'next_review': (today + timedelta(days=1)).isoformat(),
        'weekly_review': (today + timedelta(days=7)).isoformat(),
        'focus_topics': counts.most_common(3),
    }

print(build_review_plan([1, 2, 3, 40, 79], ['graphs', 'dp', 'graphs', 'tries']))

````

## Page 514: Binary search timeline

**Chapter:** 33. Visual Algorithm Atlas  
**Kind:** `search`  

Beginner goal: name the input, state, invariant, first transition, and stopping rule before reading the solution.

Source-note links: MIT 6.006 course overview, CP-Algorithms reference index, Python data structures tutorial.

````python
# Page 514: Binary search timeline
def binary_search_timeline_p514(values, target):
    lo, hi, log = 0, len(values), []
    while lo < hi:
        mid = (lo + hi) // 2
        log.append((lo, mid, hi, values[mid]))
        if values[mid] < target:
            lo = mid + 1
        else:
            hi = mid
    return {'index': lo, 'found': lo < len(values) and values[lo] == target, 'trace': log}

print(binary_search_timeline_p514([1, 4, 6, 9, 13], 9))
````

## Page 517: Queue frontier for BFS

**Chapter:** 33. Visual Algorithm Atlas  
**Kind:** `graph`  

Beginner goal: name the input, state, invariant, first transition, and stopping rule before reading the solution.

Source-note links: MIT 6.006 course overview, CP-Algorithms reference index, Python data structures tutorial.

````python
# Page 517: Queue frontier for BFS
from collections import deque

def queue_frontier_for_bfs_p517(graph, start):
    queue = deque([start])
    seen = {start}
    order = []
    while queue:
        node = queue.popleft()
        order.append(node)
        for nxt in graph.get(node, []):
            if nxt not in seen:
                seen.add(nxt)
                queue.append(nxt)
    return order

sample = {'A': ['B', 'C'], 'B': ['D'], 'C': ['D'], 'D': []}
print(queue_frontier_for_bfs_p517(sample, 'A'))
````

## Page 532: Track D: graphs without fear

**Chapter:** 34. Interactive Practice Tracks  
**Kind:** `graph`  

Beginner goal: name the input, state, invariant, first transition, and stopping rule before reading the solution.

Source-note links: MIT 6.006 course overview, CP-Algorithms reference index, Python data structures tutorial.

````python
# Page 532: Track D: graphs without fear
from collections import deque

def track_d_graphs_without_fear_p532(graph, start):
    queue = deque([start])
    seen = {start}
    order = []
    while queue:
        node = queue.popleft()
        order.append(node)
        for nxt in graph.get(node, []):
            if nxt not in seen:
                seen.add(nxt)
                queue.append(nxt)
    return order

sample = {'A': ['B', 'C'], 'B': ['D'], 'C': ['D'], 'D': []}
print(track_d_graphs_without_fear_p532(sample, 'A'))
````

## Page 552: Capstone path: personal search engine

**Chapter:** 36. Capstone Assessments and Project Rubrics  
**Kind:** `search`  

Beginner goal: name the input, state, invariant, first transition, and stopping rule before reading the solution.

Source-note links: MIT 6.006 course overview, CP-Algorithms reference index, Python data structures tutorial.

````python
# Page 552: Capstone path: personal search engine
def capstone_path_personal_search_engine_p552(values, target):
    lo, hi, log = 0, len(values), []
    while lo < hi:
        mid = (lo + hi) // 2
        log.append((lo, mid, hi, values[mid]))
        if values[mid] < target:
            lo = mid + 1
        else:
            hi = mid
    return {'index': lo, 'found': lo < len(values) and values[lo] == target, 'trace': log}

print(capstone_path_personal_search_engine_p552([1, 4, 6, 9, 13], 9))
````

## Page 553: Capstone path: route planner

**Chapter:** 36. Capstone Assessments and Project Rubrics  
**Kind:** `shortest`  

Beginner goal: name the input, state, invariant, first transition, and stopping rule before reading the solution.

Source-note links: MIT 6.006 course overview, CP-Algorithms reference index, Python data structures tutorial.

````python
# Page 553: Capstone path: route planner
from collections import deque

def capstone_path_route_planner_p553(graph, start):
    queue = deque([start])
    seen = {start}
    order = []
    while queue:
        node = queue.popleft()
        order.append(node)
        for nxt in graph.get(node, []):
            if nxt not in seen:
                seen.add(nxt)
                queue.append(nxt)
    return order

sample = {'A': ['B', 'C'], 'B': ['D'], 'C': ['D'], 'D': []}
print(capstone_path_route_planner_p553(sample, 'A'))
````

## Page 561: References and citation map

This notebook companion follows the HTML references page. Keep source notes near code: Python standard library docs for Python APIs, MDN for browser behavior, GitHub Docs for GitHub Pages, W3C WAI for accessibility, and nbformat for notebook structure.

````python
# Page 561: References and citation map
REFERENCE_RULES = {
    "API behavior": "Use official language or browser documentation.",
    "algorithm idea": "Use the course explanation first, then a trusted algorithm reference.",
    "publishing workflow": "Use GitHub Pages documentation because deployment details change.",
    "accessibility check": "Use W3C WAI guidance and test with keyboard, contrast, and labels.",
}

def choose_source(question: str) -> str:
    q = question.lower()
    if any(word in q for word in ["heap", "deque", "bisect", "cache", "sort"]):
        return REFERENCE_RULES["API behavior"]
    if any(word in q for word in ["publish", "github", "pages", "deploy"]):
        return REFERENCE_RULES["publishing workflow"]
    if any(word in q for word in ["screen reader", "keyboard", "contrast", "label"]):
        return REFERENCE_RULES["accessibility check"]
    return REFERENCE_RULES["algorithm idea"]

print(choose_source("Why does this example use a heap for the next item?"))
````
